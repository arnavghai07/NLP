```python
import nltk
from nltk.util import ngrams
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
corpus_line = "Natural language models help computers understand human communication."

corpus_tokens = word_tokenize(corpus_line)

print("Corpus Tokens:")
print(corpus_tokens)
```

```python
unigram_set = list(ngrams(corpus_tokens, 1))

print("Unigrams:")
print(unigram_set)
```

```python
bigram_set = list(ngrams(corpus_tokens, 2))

print("Bigrams:")
print(bigram_set)
```

```python
trigram_set = list(ngrams(corpus_tokens, 3))

print("Trigrams:")
print(trigram_set)
```

```python
print("Total Unigrams:", len(unigram_set))
print("Total Bigrams:", len(bigram_set))
print("Total Trigrams:", len(trigram_set))

print("\nUnigrams:", unigram_set)
print("\nBigrams:", bigram_set)
print("\nTrigrams:", trigram_set)
```
