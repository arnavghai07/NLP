```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
document_alpha = "Data science helps organizations discover useful patterns."
document_beta = "Organizations use data analysis to discover valuable patterns."

text_pair = [document_alpha, document_beta]

print("Document 1:")
print(document_alpha)

print("\nDocument 2:")
print(document_beta)
```

```python
similarity_vectorizer = TfidfVectorizer()

tfidf_vectors = similarity_vectorizer.fit_transform(text_pair)

print("TF-IDF Vectors:")
print(tfidf_vectors.toarray())
```

```python
similarity_score = cosine_similarity(tfidf_vectors[0:1], tfidf_vectors[1:2])

print("Cosine Similarity:")
print(similarity_score)
```

```python
similarity_value = similarity_score[0][0]

print("Similarity Score:", round(similarity_value, 4))
```
