```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
sentence_collection = [
    "Online classes make learning more flexible",
    "Interactive lessons make learning interesting",
    "Students enjoy flexible online lessons"
]

print("Documents:")
for line in sentence_collection:
    print(line)
```

```python
bow_converter = CountVectorizer()

bow_matrix = bow_converter.fit_transform(sentence_collection)
```

```python
feature_terms = bow_converter.get_feature_names_out()

print("Vocabulary:")
print(feature_terms)
```

```python
vector_representation = bow_matrix.toarray()

print("Bag-of-Words Representation:")
print(vector_representation)
```

```python
for document_number, vector_values in enumerate(vector_representation, start=1):
    print("Document", document_number, ":", vector_values)
```
