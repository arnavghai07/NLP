```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
text_samples = [
    "Cloud computing provides scalable storage solutions",
    "Cloud platforms offer flexible computing services",
    "Modern applications use cloud services for scalability"
]

print("Documents:")
for item in text_samples:
    print(item)
```

```python
frequency_encoder = CountVectorizer()

bow_output = frequency_encoder.fit_transform(text_samples)

print("BoW Vocabulary:")
print(frequency_encoder.get_feature_names_out())

print("\nBoW Representation:")
print(bow_output.toarray())
```

```python
importance_encoder = TfidfVectorizer()

tfidf_output = importance_encoder.fit_transform(text_samples)

print("TF-IDF Vocabulary:")
print(importance_encoder.get_feature_names_out())

print("\nTF-IDF Representation:")
print(tfidf_output.toarray())
```

```python
print("Comparison of BoW and TF-IDF")

print("\nBoW:")
print(bow_output.toarray())

print("\nTF-IDF:")
print(tfidf_output.toarray())
```
